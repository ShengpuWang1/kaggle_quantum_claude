Inspect saved verification results. See `../README.md` for commands and limitations. Raw task scores are costs, not leaderboard scores.

```python
from pathlib import Path
import json
import pandas as pd
import kickmix
ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
print('kickmix:', kickmix.__version__)
print('Workspace:', ROOT)
```

```python
rows = []
for filename in ['invsqrt-smoke.json', 'dqi-smoke-4096.json', 'dfthcsos.json']:
    result = json.loads((ROOT / 'runs/baseline' / filename).read_text())
    rows.append({'check': filename, **{k: result.get(k) for k in ['status', 'verified_shots', 'score', 'num_qubits']}})
pd.DataFrame(rows)
```
